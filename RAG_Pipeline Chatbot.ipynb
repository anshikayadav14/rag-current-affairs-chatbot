```python
import os

# Load files
data_path = '/content/drive/MyDrive/RAG_Chatbot/wikipedia_data'

all_texts = []

for filename in os.listdir(data_path):
    if filename.endswith('.txt'):
        filepath = os.path.join(data_path, filename)
        with open(filepath, 'r', encoding='utf-8') as f:
            content = f.read()
            all_texts.append({
                'filename': filename,
                'content': content
            })
        print(f"✅ Loaded: {filename} — {len(content)} characters")

print(f"\nTotal files loaded: {len(all_texts)}")
```

```python
import chromadb
import os

# Create ChromaDB folder in Drive
chroma_path = '/content/drive/MyDrive/RAG_Chatbot/chroma_db'
os.makedirs(chroma_path, exist_ok=True)

# Create ChromaDB client
client = chromadb.PersistentClient(path=chroma_path)

# Create a collection (like a table in database)
collection = client.get_or_create_collection(
    name="wikipedia_current_affairs",
    metadata={"hnsw:space": "cosine"}
)

print("✅ ChromaDB created!")
print(f"✅ Saved at: {chroma_path}")
print(f"✅ Collection name: wikipedia_current_affairs")
```

```python
import re, hashlib
from collections import Counter

print("Files:", len(all_texts))
for item in all_texts:
    print(" ", item['filename'], len(item['content']))

DATE_ONLY = re.compile(r"^(\d{1,2} [A-Z][a-z]+)\s*[–-]\s*$")
DATE_LINE = re.compile(r"^(\d{1,2} [A-Z][a-z]+)\s*[–-]\s+\S")

def clean_events(text):
    out, date = [], None
    for line in [l.strip() for l in text.split("\n") if l.strip()]:
        m = DATE_ONLY.match(line)
        if m:
            date = m.group(1)
            continue
        if DATE_LINE.match(line):
            date = None
            out.append(line)
        elif date and len(line) > 25:
            out.append(f"{date} – {line}")
        elif len(line) > 40:
            out.append(line)
        else:
            date = None
    return "\n".join(out)

def split_into_chunks(text, title, max_chars=350):
    lines = [l for l in text.split("\n") if l.strip()]
    return [f"[{title}]\n{line}" for line in lines]

hash_of = {i['filename']: hashlib.md5(i['content'].encode()).hexdigest() for i in all_texts}
counts = Counter(hash_of.values())

all_chunks, all_metadata, seen = [], [], set()
for item in all_texts:
    h = hash_of[item['filename']]
    if h in seen:
        print("Skipping duplicate:", item['filename'])
        continue
    seen.add(h)
    title = "2026 events" if counts[h] > 1 else item['filename'].replace('.txt', '').replace('_', ' ')
    cleaned = clean_events(item['content'])
    for i, chunk in enumerate(split_into_chunks(cleaned, title)):
        all_chunks.append(chunk)
        all_metadata.append({'source': item['filename'], 'chunk_id': i})
print("Total chunks:", len(all_chunks))

client.delete_collection("wikipedia_current_affairs")
collection = client.get_or_create_collection(
    name="wikipedia_current_affairs", metadata={"hnsw:space": "cosine"}
)
for i in range(0, len(all_chunks), 100):
    collection.add(
        documents=all_chunks[i:i+100],
        metadatas=all_metadata[i:i+100],
        ids=[f"chunk_{j}" for j in range(i, min(i + 100, len(all_chunks)))]
    )
print("Total in database:", collection.count())
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

print("Loading TinyLlama...")

tokenizer = AutoTokenizer.from_pretrained(
    "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
)

model = AutoModelForCausalLM.from_pretrained(
    "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    torch_dtype=torch.float16,
    device_map="auto"
)

print("✅ TinyLlama loaded!")
```

```python
# Check if model loaded properly
print(f"Model type: {type(model)}")
print(f"Tokenizer type: {type(tokenizer)}")

# Quick test
test_input = tokenizer("Hello", return_tensors="pt")
print(f"\n✅ Tokenizer working!")
print(f"Input tokens: {test_input}")
```

```python
import re
from transformers import pipeline

generator = pipeline(
    "text-generation", model=model, tokenizer=tokenizer,
    max_new_tokens=200, do_sample=False,
    return_full_text=False, pad_token_id=tokenizer.eos_token_id
)

def trim_to_sentence(text):
    end = max(text.rfind("."), text.rfind("!"), text.rfind("?"))
    return text[:end + 1] if end != -1 else text

def clean_answer(text):
    stop_phrases = [
        "the context does not", "context does not contain",
        "i don't know", "not mentioned in the context",
        "no information", "does not provide"
    ]
    lower = text.lower()
    cut = len(text)
    for p in stop_phrases:
        idx = lower.find(p)
        if idx != -1:
            cut = min(cut, idx)
    return text[:cut].strip()

# Tightened from 8 -> 4: fewer chunks means less chance of the model
# blending facts from two unrelated chunks into one false claim.
N_RESULTS = 4

# Tightened from 0.9 -> 0.6: 0.9 was letting through chunks that just
# shared a keyword (e.g. "Haryana") with the question but weren't
# actually relevant. Re-test this value against your own question set
# if you change embedding models or chunking.
DISTANCE_THRESHOLD = 0.65

def rag_answer(question):
    results = collection.query(query_texts=[question], n_results=N_RESULTS)

    docs = results['documents'][0]
    distances = results['distances'][0]

    if not docs or distances[0] > DISTANCE_THRESHOLD:
        return "I don't know"

    # Only keep chunks reasonably close to the best match, not just
    # under the absolute threshold - stops a strong top hit from
    # dragging in weakly related chunks that get blended into the answer.
    kept = []
    for doc, dist in zip(docs, distances):
        if dist <= distances[0] + 0.15:
            if doc.startswith("[") and "\n" in doc:
                doc = doc.split("\n", 1)[1]
            kept.append(doc)

    context = "\n\n".join(kept)[:2500]

    messages = [
        {"role": "system", "content": "You are a helpful assistant that answers questions using the given context."},
        {"role": "user", "content": f"Context:\n{context}\n\nQuestion: {question}\n\nAnswer the question in 2-3 sentences using only the context above. If the context does not contain the answer, say: I don't know."},
    ]
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    answer = generator(prompt)[0]['generated_text'].strip()
    answer = re.sub(r"^\d+\.\s*", "", answer)
    answer = clean_answer(answer)
    return answer

print("✅ rag_answer updated: n_results=4, distance_threshold=0.6, return statement fixed")
```

```python
import gradio as gr

gr.close_all()

def trim_to_sentence(text):
    if not text:
        return "I don't know"
    end = max(text.rfind("."), text.rfind("!"), text.rfind("?"))
    return text[:end + 1] if end != -1 else text

def chat(message, history):
    try:
        answer = rag_answer(message)
    except Exception as e:
        return f"Error occurred: {e}"
    return trim_to_sentence(answer)

gr.ChatInterface(
    chat,
    title="Current Affairs RAG Bot",
    description="Ask about India in 2026",
    examples=["Tell me about IPL 2026", "What happened in India in January 2026?"]
).launch(share=True, debug=True)
```
